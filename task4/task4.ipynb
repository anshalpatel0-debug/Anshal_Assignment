```python
from huggingface_hub import login
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

# 1. Login (replace with your token)
#login(token="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")

# 2. Load model
m = "Qwen/Qwen2.5-3B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(m)
model = AutoModelForCausalLM.from_pretrained(m)
chatbot = pipeline("text-generation", model=model, tokenizer=tokenizer)

# ==========================================
# CHAIN-OF-THOUGHT (CoT) PROMPT FOR TASK 4
# ==========================================
cot_prompt = """
You are a clinical decision support assistant. Analyze the given patient presentation by breaking down your reasoning step-by-step before providing a final recommendation.

Follow these explicit reasoning steps:
Step 1: Identify key symptoms and vital signs mentioned.
Step 2: Determine if there are any immediate red-flag indicators or severe risks.
Step 3: Evaluate potential urgency level (Low, Moderate, High, Emergency).
Step 4: Formulate a final safe recommendation.

Patient Case:
"A 65-year-old patient reports sudden onset of severe chest pressure, shortness of breath, and diaphoresis (excessive sweating) lasting for 30 minutes."

Begin your step-by-step reasoning now:
"""

messages = [{"role": "user", "content": cot_prompt}]
prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

print("Executing Chain-of-Thought Analysis...\n")
response = chatbot(prompt, max_new_tokens=300)

# Print ONLY the generated output
full_response = response[0]['generated_text']
print(full_response[len(prompt):].strip())
```
