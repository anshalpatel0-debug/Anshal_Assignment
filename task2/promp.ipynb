```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

# Load model 
m = "Qwen/Qwen2.5-3B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(m)

model = AutoModelForCausalLM.from_pretrained(m)

chatbot = pipeline("text-generation", model=model, tokenizer=tokenizer)

structured_prompt = """
[ROLE]
You are a Board-Certified Clinical Specialist in Endocrinology working as an AI Healthcare Advisor.

[CONTEXT]
A patient's caregiver is seeking an accurate, easy-to-understand explanation of Diabetes Mellitus.

[CONSTRAINTS]
- Do NOT provide personal medical diagnoses or prescribe specific medication dosages.
- Maintain an empathetic, clear, and professional clinical tone.
- Avoid overly dense medical jargon without providing a simple explanation.

[OUTPUT FORMATTING]
Format your response using clean Markdown strictly adhering to these headings:

## Disease Overview
Briefly explain what Diabetes Mellitus is and its main types.

## Symptoms
List common physical signs and symptoms using bullet points.

## Risk Factors
List lifestyle, genetic, and environmental risk factors using bullet points.

## Prevention
Provide actionable steps for prevention and lifestyle management.

## When to Consult a Doctor
Specify red-flag symptoms requiring immediate medical evaluation.

[USER QUERY]
Tell me about diabetes.
"""

messages = [{"role": "user", "content": structured_prompt}]
prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

print("Generating structured Markdown output...\n")
response = chatbot(prompt, max_new_tokens=500)
full_response = response[0]['generated_text']
print(full_response[len(prompt):].strip())
```
