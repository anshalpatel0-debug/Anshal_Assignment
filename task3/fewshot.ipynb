```python
from huggingface_hub import login
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

login(token="")

m = "Qwen/Qwen2.5-3B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(m)
model = AutoModelForCausalLM.from_pretrained(m)
chatbot = pipeline("text-generation", model=model, tokenizer=tokenizer)


few_shot_prompt = """
You are a medical assistant. Convert clinical notes into structured summaries containing Symptoms and Severity (Low, Moderate, High).

Example 1:
Input: Patient reports mild runny nose and occasional sore throat for 2 days. No fever.
Output:
Symptoms:
- Runny nose
- Sore throat
Severity:
Low

Example 2:
Input: Patient presents with persistent severe chest pain radiating to the left arm and heavy shortness of breath.
Output:
Symptoms:
- Severe chest pain
- Radiation of pain to left arm
- Shortness of breath
Severity:
High

Example 3:
Input: Patient has had a dry cough and a temperature of 101°F for 3 days. Experiencing fatigue.
Output:
Symptoms:
- Dry cough
- Fever (101°F)
- Fatigue
Severity:
Moderate

Now complete the following:

Input: Patient has cough and fever.
Output:
"""

messages = [{"role": "user", "content": few_shot_prompt}]
prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

print("Running Few-Shot Extraction...\n")
response = chatbot(prompt, max_new_tokens=150)

# Print ONLY the generated answer
full_response = response[0]['generated_text']
print(full_response[len(prompt):].strip())
```
